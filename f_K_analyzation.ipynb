```python
scale = len(f_Kp2_reload[0].parent().gens())
dimension = len(f_Kp2_reload)
X = FormalVariables(scale)
```

```python
# analyzation part
from sage.rings.polynomial.polydict import ETuple
deg = 0
monomi = ETuple([1 if i < deg else 0 for i in range(scale)])
res_list = []

for n in range(max(deg-1,0),dimension):
    # modification to f_Kp2
    modif = (-1)^n * math.factorial(n+1-deg)
    nmlzd = f_Kp2_reload[n]/modif
    res = nmlzd.dict().get(monomi,0)
    res_list.append(res)
    print(f'coefficient of {n+1}-th normalized f_Kp2 at {prod(X.variables()[:deg])} is {res}')

# res_list
print('\n',res_list)
```
